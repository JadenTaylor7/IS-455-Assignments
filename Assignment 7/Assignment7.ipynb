## Part 2: Examine the data and parse it

```python
for jobType in listOfJobs:
    url = f"https://data.usajobs.gov/api/Search?Keyword={jobType}&RemunerationMinimumAmount={minimumSalary}&RemunerationMaximumAmount={maximumSalary}"
    response = requests.get(url, headers=someHeader)

    if response.status_code == 200:
        data = response.json()

        dataObject = json.dumps(data, indent= 4) #for seeing what we are working with
        with open("test.json", "a") as outfile:
            outfile.write(f"\n\nJob Type: {jobType}\n\n")
            outfile.write(dataObject)


        # Process the JSON data
        for item in data["SearchResult"]["SearchResultItems"]:
            descriptor = item["MatchedObjectDescriptor"]
            df.loc[len(df)] = [descriptor["PositionTitle"],
                               descriptor["PositionRemuneration"][0]["MinimumRange"],
                               descriptor["PositionRemuneration"][0]["MaximumRange"],
                               descriptor["PositionURI"]]
    else:
        print(f"{jobType} was done messed up, A-aron Status code: {response.status_code}")



# Time to style
pd.set_option('display.max_columns', None)  # allow all columns to be seen

print(df.head())
#print(df)
df.to_csv("Cool Jobs.csv")


# code below is old code for a single search
# import pandas as pd

# dataObject = json.dumps(data, indent= 4) #for seeing what we are working with
# with open("test.json", "w") as outfile:
#     outfile.write(dataObject)

# df = pd.DataFrame(columns=["PositionTitle", "MinimumRange", "MaximumRange", "PositionURI"])

# for item in data["SearchResult"]["SearchResultItems"]:
#     descriptor = item["MatchedObjectDescriptor"]
#     df.loc[len(df)] = [descriptor["PositionTitle"], 
#                        descriptor["PositionRemuneration"][0]["MinimumRange"], 
#                        descriptor["PositionRemuneration"][0]["MaximumRange"], 
#                        descriptor["PositionURI"]]


# # Time to style
# pd.set_option('display.max_columns', None) #allow all columns to bee seen

# print(df.head())
# print(df)
```
