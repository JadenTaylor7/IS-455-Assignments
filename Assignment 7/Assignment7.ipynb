## Part 2: Examine the data and parse it

```python
import pandas as pd

dataObject = json.dumps(data, indent= 4) #for seeing what we are working with
with open("test.json", "w") as outfile:
    outfile.write(dataObject)


df = pd.DataFrame(columns=["PositionTitle", "MinimumRange", "MaximumRange", "PositionURI"])

for item in data["SearchResult"]["SearchResultItems"]:
    descriptor = item["MatchedObjectDescriptor"]
    df.loc[len(df)] = [descriptor["PositionTitle"], 
                       descriptor["PositionRemuneration"][0]["MinimumRange"], 
                       descriptor["PositionRemuneration"][0]["MaximumRange"], 
                       descriptor["PositionURI"]]


# Time to style
pd.set_option('display.max_columns', None) #allow all columns to bee seen

print(df.head())
print(df)
```
